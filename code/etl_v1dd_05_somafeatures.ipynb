# Read V1DD soma features parquet

Reads and displays `soma_features.parquet` from the V1DD 1196 release.

```python
from pathlib import Path

import polars as pl
```

```python
DATA_ROOT = Path("/data/v1dd_v1196_20260827/v1dd_v1196_soma_features")
SOMA_FEATURES_PATH = DATA_ROOT / "soma_features.parquet"
```

```python
soma_features = pl.read_parquet(SOMA_FEATURES_PATH)
print(soma_features.shape)
```

## Read the same table via CAVE client

Table name `soma_features` comes from the shared path `.../tables/soma_features.parquet`.

```python
import os

import caveclient
```

```python
cave_token = os.environ["CUSTOM_KEY_2"]
mat_version = 1196
```

```python
client = caveclient.CAVEclient(
    "v1dd",
    server_address="https://global.em.brain.allentech.org",
    auth_token=cave_token,
    version=mat_version,
)
```

```python
client.materialize.get_tables()
```

```python
client.materialize.get_views()
```

```python
soma_features_cave = client.materialize.query_table("cell_type_multifeature_v1")
print(soma_features_cave.shape)
```
