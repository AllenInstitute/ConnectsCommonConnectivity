```python
import polars as pl
from matplotlib import pyplot as plt
import numpy as np
import umap
from sklearn.preprocessing import StandardScaler
from utils import adjacencyplot, cell_type_palette
import seaborn as sns
import os
```

```python
datalake_root = "/data/microns1412/"
```

```python
conn_df = pl.read_delta(os.path.join(datalake_root, "cellcellconnectivitylong/"))
result = conn_df.filter(pl.col("measurement_type") == "SUM_ANATOMICAL_SIZE")

datasets = pl.read_delta(os.path.join(datalake_root, "dataset/"))
datasets
```

```python
result.to_pandas()
```

```python
data_items = pl.read_delta(os.path.join(datalake_root, "dataitem_dataset_association"))
prf_cells=data_items.filter(pl.col("dataset_id") == "minnie65_v1412_proofread")
feat_cells= data_items.filter(pl.col("dataset_id") == "minnie65_v1412_csm_cluster")
```

```python
clusters=pl.read_delta(os.path.join(datalake_root, "cluster"))
cluster_members =pl.read_delta(os.path.join(datalake_root, "clustermembership"))
cell_types=cluster_members.join(clusters, left_on="cluster", right_on="id").filter(pl.col("level") == 2)
```

```python
cell_types
```

```python
cell_features = pl.read_delta(os.path.join(datalake_root, "cellfeatures/csm_cluster_features"))
nodes=cell_types.select(["item", "cluster", "parent"]).join(cell_features, left_on="item", right_on="id")
```

```python
cell_feature_sets =  pl.read_delta(os.path.join(datalake_root, "cellfeatureset"))
feature_set = cell_feature_sets.filter(pl.col("id") == "csm_cluster_features")
features =  pl.read_delta(os.path.join(datalake_root, "cellfeaturedefinition")).filter(pl.col('id').is_in(feature_set['feature_definition_ids']))
features
```

```python
cell_type_palette={}
for row in clusters.rows():
    cell_type_palette[row[0]]=row[5]
```

```python
df=result.to_pandas()
node_df = nodes.to_pandas()
df['log_syanpse_size'] = np.log(df.value)
f, ax = plt.subplots(figsize=(8,24))
adjacencyplot(df,
              node_df,
              node_id_col="item",
              row_groupby=["parent", "cluster"],
              col_groupby=["parent", "cluster"],
              col_sortby=["median_density"],
              node_palette=cell_type_palette,
              pre_id_col="postsynaptic_cell",
              post_id_col="presynaptic_cell",
              arc_labels = None,
              weight_col="log_syanpse_size",
              plot_type="scattermap",
              row_value_bars=["soma_volume_um","median_density"],
              col_value_bars=["soma_volume_um","median_density"],
              edge_hue=False,
              hue_norm=(1,10),
              sizes = (.1, 2),
              size_norm=(0,13),
              ax=ax)
```

```python
umap = pl.read_delta(os.path.join(datalake_root, "cellfeatures/csm_cluster_features_umap")).drop(['project_id','feature_set_id'])
cell_cluster =  pl.read_delta(os.path.join(datalake_root, "clustermembership")).select(['item', 'cluster', 'parent'])
anatomy = pl.read_delta(os.path.join(datalake_root, "cellfeatures/minnie65_std_transform_coordinates")).drop(['project_id','feature_set_id'])
nodes=cell_cluster.join(umap, left_on='item', right_on='id').join(anatomy, left_on='item', right_on='id')

pallete = {}
for row in clusters.select(['id', 'hex_color']).rows():
    pallete[row[0]]=row[1]
```

```python
nodes
```

```python
f, ax = plt.subplots(figsize=(12,12))

sns.scatterplot(data=nodes,
                x='x_umap',
                y='y_umap',
                hue='cluster',
                s=5,
                palette=pallete, 
                linewidths=0, ax=ax)
```
